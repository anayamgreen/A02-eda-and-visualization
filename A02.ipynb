```python
# Your Phase 2 solution to the problem goes here.
# did not use any critiques
 import pandas as pd
import matplotlib.pyplot as plt

# Make sure Foreign Gift Amount is numeric
df['Foreign Gift Amount'] = pd.to_numeric(
    df['Foreign Gift Amount'],
    errors='coerce'
)

# -----------------------------------
# 1. TOP 15 COUNTRIES BY NUMBER OF GIFTS
# -----------------------------------

top_15_gifts = (
    df.groupby('Country of Giftor')
      .size()
      .reset_index(name='Number of Gifts')
      .sort_values('Number of Gifts', ascending=False)
      .head(15)
)

print("Top 15 Countries by Number of Gifts")
print(top_15_gifts)


# -----------------------------------
# 2. TOP 15 COUNTRIES BY TOTAL AMOUNT GIVEN
# -----------------------------------

top_15_amount = (
    df.groupby('Country of Giftor')['Foreign Gift Amount']
      .sum()
      .reset_index()
      .rename(columns={'Foreign Gift Amount': 'Total Amount Given'})
      .sort_values('Total Amount Given', ascending=False)
      .head(15)
)

print("\nTop 15 Countries by Total Amount Given")
print(top_15_amount)


# -----------------------------------
# 3. OPTIONAL: DISPLAY AS BAR CHARTS
# -----------------------------------

# Top 15 by number of gifts
plt.figure(figsize=(12, 6))

plt.bar(
    top_15_gifts['Country of Giftor'],
    top_15_gifts['Number of Gifts']
)

plt.xticks(rotation=45, ha='right')
plt.title('Top 15 Countries by Number of Gifts')
plt.xlabel('Country')
plt.ylabel('Number of Gifts')
plt.tight_layout()
plt.show()


# Top 15 by amount given
plt.figure(figsize=(12, 6))

plt.bar(
    top_15_amount['Country of Giftor'],
    top_15_amount['Total Amount Given']
)

plt.xticks(rotation=45, ha='right')
plt.title('Top 15 Countries by Total Amount Given')
plt.xlabel('Country')
plt.ylabel('Total Amount ($)')
plt.tight_layout()
plt.show()
```
